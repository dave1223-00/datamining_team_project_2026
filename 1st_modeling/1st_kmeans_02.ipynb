단순 피쳐 갯수 축소

```python
import mglearn
import pandas as pd
from sklearn.cluster import KMeans
```

```python
from sklearn.metrics import silhouette_score
```

```python
df = pd.read_csv('single_variable_data.csv')
```

```python
df.head()
```

```python

K_range = range(2,14)
    
for i in range(0,24):
    print(f"{i}번째 열 삭제")
    declined_columns_df = df.drop(df.columns[i], axis=1)
    inertia_kmeans = []
    sil_scores_kmeans = []
    for n_clusters in K_range:
        kmeans = KMeans(n_clusters=n_clusters, random_state=42)
        # n_init: initial centroid 변경횟수   , n_init=10
        # 없이 했다가 추가하니까 실루엣스코어 살짝씩 올랐음
        cluster_labels = kmeans.fit_predict(declined_columns_df)
    
        inertia_kmeans.append(kmeans.inertia_)
        sil_scores_kmeans.append(silhouette_score(declined_columns_df, cluster_labels))
    print("<inertia>")
    print(inertia_kmeans)
    print("<silhouette socre>")
    print(sil_scores_kmeans)
    
```

```python
K_range = range(2,14)

#열 1개 삭제

for i in range(0,24):
    print(f"{i}번째 열 삭제")
    declined_columns_df = df.drop(df.columns[i], axis=1)
    
    highest_sil=0
    for n_clusters in K_range:
        kmeans = KMeans(n_clusters=n_clusters, random_state=42)
        # n_init: initial centroid 변경횟수   , n_init=10
        # 없이 했다가 추가하니까 실루엣스코어 살짝씩 올랐음
        cluster_labels = kmeans.fit_predict(declined_columns_df)
        
        if (silhouette_score(declined_columns_df, cluster_labels) > highest_sil):
            highest_sil = silhouette_score(declined_columns_df, cluster_labels)
            
    print(f"highest_sil: {highest_sil}")
```

```python
K_range = range(2,14)

#열 2개 삭제

for j in range(0,24):
    range_without_j = [x for x in range(24) if x != j]
    for i in range_without_j:
        print(f"{j}, {i}번째 열 삭제")
        declined_columns_df = df.drop([df.columns[i], df.columns[j]], axis=1)

        highest_sil=0
        
        for n_clusters in K_range:
            kmeans = KMeans(n_clusters=n_clusters, random_state=42)
       
            cluster_labels = kmeans.fit_predict(declined_columns_df)
        
            if (silhouette_score(declined_columns_df, cluster_labels) > highest_sil):
                highest_sil = silhouette_score(declined_columns_df, cluster_labels)
            
        print(f"highest_sil: {highest_sil}")
```

```python
K_range = range(2,14)

#열 4개 삭제
for l in range(0,24):
for k in range(0,24): 
    range_without_k = [x for x in range(24) if x != k]
    for j in range(0,24):
        range_without_j = [x for x in range(24) if x != j]
        for i in range_without_j:
            print(f"{j}, {i}번째 열 삭제")
            declined_columns_df = df.drop([df.columns[i], df.columns[j], df.columns[k]], axis=1)

            highest_sil=0
        
            for n_clusters in K_range:
                kmeans = KMeans(n_clusters=n_clusters, random_state=42)
       
                cluster_labels = kmeans.fit_predict(declined_columns_df)
        
                if (silhouette_score(declined_columns_df, cluster_labels) > highest_sil):
                    highest_sil = silhouette_score(declined_columns_df, cluster_labels)
            
            print(f"highest_sil: {highest_sil}")
```
